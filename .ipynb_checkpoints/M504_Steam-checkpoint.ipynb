
# Steam Market Intelligence Report
**Client:** A new indie game studio planning to release its first game on Steam.  
**Project goal:** Use historical Steam data to recommend a good launching stratergy for a company looking to launch their game on steam.

## Executive Summary

This report looks at Steam marketplace data which is scraped from official steam webiste.

**Key findings:**
- Games priced under $15 tend to reach the most players on Steam.
- Games that consistently earns user reviews tend to have higher estimated ownership numbers.
- Windows ecosystem represents over 79% of all Steam games.
- Competition on Steam has exploded from about 25 times what it was in 2013, this makes it harder for new games to get attention.
- Most Games fall under 20,000 estimated owners category, this marks how important it is to set realistic launch targets.

**Strategic recommendation:** The studio should aim a low-risk and accessibility-focused launch strategy which implies affordable pricing, Windows-first development, review generation incentive, and loyal community making. Expectations should be benchmarked against realistic targets.

- Dataset: Steam games market dataset from Kaggle.
- Source URL: https://www.kaggle.com/datasets/fronkongames/steam-games-dataset
- Client: A new indie studio seeking guidance on pricing, platform support, reviews, and launch strategy.

## Methodology

Data Acquisition → Preprocessing → EDA → Business Analysis → Recommendations.

Tools: Python 3, Pandas, NumPy, and Plotly Express. 

All analysis was conducted in a reproducible Jupyter Notebook environment.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio


pd.options.display.max_columns = None
pio.renderers.default = "svg"
```

## Data Loading and Initial Validation

Assigining the correct header to the dataset and counting the number of rows.

```python
cols = [
    "AppID", "Name", "Release date", "Estimated owners", "Peak CCU",
    "Required age", "Price", "Discount", "DLC count", "About the game",
    "Supported languages", "Full audio languages", "Reviews", "Header image",
    "Website", "Support url", "Support email", "Windows", "Mac", "Linux",
    "Metacritic score", "Metacritic url", "User score", "Positive", "Negative",
    "Score rank", "Achievements", "Recommendations", "Notes",
    "Average playtime forever", "Average playtime two weeks",
    "Median playtime forever", "Median playtime two weeks", "Developers",
    "Publishers", "Categories", "Genres", "Tags", "Screenshots", "Movies"
]

df = pd.read_csv('games.csv', header=None, names=cols, skiprows=1, engine='python')
print('rows:', len(df))
print('columns:', len(df.columns))
df.head(3)
```

## Dataset Description

This data set is created by scrapping the records from original steam website. It contains info such as pricing, reviews, hours played by users of the game, launch date, etc this info will help us make a stratergy for a company looking forward to launch thier game on steam.

```python
description = {
    'Games': len(df),
    'Columns': len(df.columns),
    'Unique prices': df['Price'].nunique(),
    'Estimated owners categories': df['Estimated owners'].nunique(),
    'Release years': df['Release date'].nunique(),
}
print(description)
```

## Data Quality Assessment

Finding missing values and inconsistent data formats issues that must be addressed before analysis.

```python
missing = df.isnull().sum().sort_values(ascending=False)
duplicates = df.duplicated().sum()
print('Duplicates:', duplicates)
print('Missing columns:', missing[missing > 0].head(12))
```

### Data Quality Findings

We checked all 122,611 games, no duplicates showed up. Ownership isn’t given as exact numbers; instead, it’s listed in ranges, so any analysis based on ownership is an estimate, not a precise count. Some columns, like Movies, Score Rank, Metacritic URL, and Reviews are missing a lot of data, so we left them out of the analysis. Other fields, such as Website, Support URL, Tags, and Categories, have some missing values, so we only used them when information was actually available. The key business variables Price, number of Reviews, Peak CCU, and platform support held up fine, with no major data quality problems.

### Potential Biases

Since the ownership data is grouped in broad ranges, you don’t get much detail. Games that fall inside the same range look more similar than they might actually be. There’s also survivor bias lurking in the engagement stats because only games with actual playtime get counted.

## Data Preprocessing

Convert ownership range to numeric midpoints, extract release year, compute review metrics, classify price tiers, and define platform support categories.

```python
def parse_owner_range(value):
    if isinstance(value, str) and ' - ' in value:
        low, high = value.split(' - ')
        try:
            return (int(low.replace(',', '')) + int(high.replace(',', ''))) / 2
        except ValueError:
            return np.nan
    return np.nan

df['Estimated owners midpoint'] = df['Estimated owners'].astype(str).apply(parse_owner_range)
df['Release Year'] = pd.to_datetime(df['Release date'], errors='coerce').dt.year
df['Review count'] = df['Positive'] + df['Negative']
df['Positive ratio'] = np.where(df['Review count'] > 0, df['Positive'] / df['Review count'], np.nan)

def price_category(price):
    if price == 0:
        return 'Free'
    if price < 10:
        return 'Low'
    if price < 30:
        return 'Mid'
    return 'High'

df['Price Category'] = df['Price'].apply(price_category)

def platform_support(row):
    platforms = []
    if row['Windows']:
        platforms.append('Windows')
    if row['Mac']:
        platforms.append('Mac')
    if row['Linux']:
        platforms.append('Linux')
    if len(platforms) == 0:
        return 'None'
    if len(platforms) == 1:
        return platforms[0] + ' Only'
    if len(platforms) == 2:
        return '+'.join(platforms)
    return 'All platforms'

df['Platform Support'] = df.apply(platform_support, axis=1)

print(df[['Estimated owners', 'Estimated owners midpoint', 'Price', 'Price Category', 'Platform Support', 'Positive ratio', 'Release Year']])
```

## Q1. How much is the competition in Steam marketplace?

**Business value:** Market competition alert us about discoverability, marketing investment, and timing.

```python
df["Release Year"] = pd.to_datetime(df["Release date"], errors="coerce").dt.year
df_year = df[df["Release Year"].notna()]

games_per_year = (
    df_year["Release Year"]
    .value_counts()
    .sort_index()
    .reset_index()
)
games_per_year.columns = ["Release Year", "Number of Games"]

fig = px.line(
    games_per_year,
    x="Release Year",
    y="Number of Games",
    markers=True,
    title="Trend of Game Releases on Steam Over Time"
)

fig.update_layout(
    template="plotly_white",
    height=700,
    width = 1000,
    title_x=0.5,
    xaxis_title="Release Year",
    yaxis_title="Number of Games Released",
    xaxis_showgrid = False,
    yaxis_showgrid = False
)


fig.add_vline(x=2025, line_dash="dash", line_color="rgba(255,0,0,0.5)", annotation_text="2025", annotation_position="top")
fig.add_vline(x=2019, line_dash="dash", line_color="rgba(255,0,0,0.5)", annotation_text="2019", annotation_position="top")

fig.add_annotation(
    x=2025,
    y=1,
    yref="paper",
    text="Insufficient Data",
    xshift=25,
    arrowhead=2,
    ax=30,   
    ay=-40,)

fig.update_traces(mode="lines") 
fig.update_traces(line=dict(width=3))


fig.show()
```

### Interpretation

Steam’s landscape has changed massively—back in 2013, you’d find under a thousand releases; by 2025, that number’s shot up to nearly 25,000. That’s a crush of competition. Just making a solid game isn’t enough any more.

### Business Insight

If you want your game to succeed, you need more than quality. You need people to see it. These days, that means smart marketing, building a community before launch, and sharpening your Steam store page so it actually converts browsers into wishlisters.

### Recommendation

My advice? Don’t go it alone on launch day. Build hype early. Grow a community, optimize your store page, and get that wishlist count up. Try not to launch when everyone else does avoid those crowded release windows if you can.


## Q2. What pricing strategy should the studio adopt?

**Business value:** The launch price directly affects sales and player adoption.


```python
q1_df = df[~df['Estimated owners'].isin(['0 - 0', '0 - 20000'])].copy()

bracket_order = [
    '20000 - 50000', '50000 - 100000',
    '100000 - 200000', '200000 - 500000', '500000 - 1000000',
    '1000000 - 2000000', '2000000 - 5000000',
    '5000000 - 10000000', '10000000 - 20000000', '20000000 - 50000000'
]

dist = (
    pd.crosstab(q1_df['Price Category'], q1_df['Estimated owners'], normalize='index')
    .mul(100)
    .reset_index()
)

dist_long = dist.melt(id_vars='Price Category', var_name='Owner Bracket', value_name='Percentage')
dist_long = dist_long[dist_long['Owner Bracket'].isin(bracket_order)]
dist_long['Owner Bracket'] = pd.Categorical(dist_long['Owner Bracket'], categories=bracket_order, ordered=True)
dist_long = dist_long.sort_values(['Price Category', 'Owner Bracket'])
dist_long.head()
```

```python
fig = px.bar(
    dist_long,
    x='Price Category',
    y='Percentage',
    color='Owner Bracket',
    category_orders={
        'Owner Bracket': bracket_order,
        'Price Category': ['Free', 'Low', 'Mid', 'High']
    },
    barmode='stack',
    title='Owner Bracket Distribution Within Each Price Category (excl. zero-owner games)',
    labels={'Percentage': '% of Games in Price Category', 'Price Category': 'Price Category'}
)

fig.update_layout(
    height=700,
    width = 1000,
)

fig.show()
```

**Note:** The data leaves out games with fewer than 20,000 owners to keep the focus on titles that cleared the studio’s own sales goal.

### Interpretation

Games that are either free or mid-priced pop up more often in the larger ownership brackets. High-priced games? They mostly stay at the low end.

### Business Insight

Pricing really shapes your reach. Mid-priced games at $5–15 strike a good balance, accessible enough for more buyers, yet still promising enough quality. Charging more raises the hurdle for players, making it tougher to build momentum.

### Recommendation

Aim for that sweet spot, launch in the $5–15 range and throw in an early-bird discount. That’s how you drive those all-important sales and reviews in the early days.

## Q3. Which ecosystem should the game be made for?

**Business value:** Supported platform must reach maximize audience.

```python
platform_summary = df.groupby('Platform Support', observed=True).agg(
    games=('AppID', 'size'),
    median_owners=('Estimated owners midpoint', 'median'),
    mean_positive_ratio=('Positive ratio', 'mean')
).reset_index().sort_values(by='games', ascending=False)
platform_summary = platform_summary[platform_summary["games"] > 30]
```

```python
fig = px.bar(
    platform_summary,
    x='games',
    y='Platform Support',
    orientation='h',
    text='games',
    title='Steam Games by Platform Support',
    labels={'games': 'Number of Games', 'Platform Support': 'Platform Support'}
)
fig.update_layout(yaxis={'categoryorder':'total descending'},height=500,
    width = 1000,)

fig.show()
```

### Interpretation and Business Insight

Almost 80% of Steam games only run on Windows. Games that support multiple platforms do have slightly higher review scores, but the boost isn’t dramatic.

### Recommendation

Stick with Windows at launch. That’s where you get the biggest audience for the lowest cost. See how things go, then consider Mac or Linux support only if there’s clear demand and you have resources left.

## Q4. Do we require reviews for commercial success?

**Business value:** If review sentiment impacts ownership, the studio should make review generation a priority.


```python
def format_range(label):
    parts = label.split(" - ")
    start = int(parts[0])
    end = int(parts[1])
    
    def short(n):
        if n >= 1_000_000:
            return f"{n//1_000_000}M"
        elif n >= 1_000:
            return f"{n//1_000}K"
        else:
            return str(n)
    
    return f"{short(start)} - {short(end)}"


df_reviews = df.melt(
    id_vars=["Estimated owners"],
    value_vars=["Positive", "Negative"],
    var_name="Review Type",
    value_name="Review Count"
)

df_reviews["Estimated owners"] = df_reviews["Estimated owners"].apply(format_range)

owner_order = [
    "0 - 20K",
    "20K - 50K",
    "50K - 100K",
    "100K - 200K",
    "200K - 500K",
    "500K - 1M",
    "1M - 2M",
    "2M - 5M",
    "5M - 10M",
    "10M - 20M",
    "20M - 50M",
    "50M - 100M",
    "100M - 200M",
]

box_plot = px.box(
    df_reviews,
    x="Estimated owners",
    y="Review Count",
    color="Review Type",
        color_discrete_map={
        "Positive": "green",
        "Negative": "red"
    },
    points=False,
    category_orders={
        "Estimated owners": owner_order
    }
)

box_plot.update_layout(
    template="plotly_white",
    height=700,
    width = 1500,
        legend=dict(
        orientation="h",
        y=1.05,
        x=1,
        title_text = "Type",
        xanchor="right",

    ),
    yaxis_showgrid = False,
    xaxis_tickangle=0

)

box_plot.update_yaxes(type="log")

box_plot.show()
```

### Interpretation

When games reach higher ownership tiers, they rack up way more reviews but the ratio of positive reviews stays pretty level. It means that while volume helps, sentiment matters more.


### Recommendation

Push for early reviews by keeping your community engaged and building relationships with creators. Once you launch, keep an ear out for feedback. Show you’re listening and adapting.

## Q5. Successful games comprise of which characteristics?

**Business value:** Important success factors that the company should be aware of.

```python
top_threshold = df['Estimated owners midpoint'].quantile(0.90)
top_games = df[df['Estimated owners midpoint'] >= top_threshold]
```

```python
success_summary = pd.DataFrame({
    'Metric': [
        'Median Price',
        'Median Peak CCU',
        'Median Average Playtime',
        'Median Positive Ratio',
        'Windows Share',
        'Multi-platform Share'
    ],
    'Top 10%': [
        top_games['Price'].median(),
        top_games['Peak CCU'].median(),
        top_games['Average playtime forever'].median(),
        top_games['Positive ratio'].median(),
        (top_games['Platform Support'].str.contains('Windows')).mean(),
        top_games['Platform Support'].isin(['Windows+Mac', 'Windows+Linux', 'All platforms']).mean()
    ],
    'All games': [
        df['Price'].median(),
        df['Peak CCU'].median(),
        df['Average playtime forever'].median(),
        df['Positive ratio'].median(),
        (df['Platform Support'].str.contains('Windows')).mean(),
        df['Platform Support'].isin(['Windows+Mac', 'Windows+Linux', 'All platforms']).mean()
    ]
})
success_summary
```

```python
top_games[['Price', 'Peak CCU', 'Review count', 'Positive ratio', 'Average playtime forever']].describe().round(2)
```

```python
fig = px.bar(
    success_summary.melt(id_vars=['Metric'], var_name='Group', value_name='Value'),
    x='Value',
    y='Metric',
    color='Group',
    barmode='group',
    title='Top 10% Games vs Entire Catalog',

)
fig.update_layout(
    height=700,
    width = 1000,
)
fig.show()
```

### Interpretation and Business Insight

Top-performing games all share one thing: engagement. They keep players playing session times and peak concurrent users are much higher for hits, even when review scores stay consistent across the board..

### Recommendation

If you want your game to take off, focus on player retention just as much as acquisition. Build features with replay value. Then, once you launch, keep a close eye on playtime and engagement for the first month.

## Q6. Which factors are showing the strongest relationship with ownership?

**Business value:** Understanding which game metric correlate strongly with ownership.


```python
correlation_cols = [
    'Price',
    'Peak CCU',
    'Review count',
    'Positive ratio',
    'Average playtime forever',
    'Estimated owners midpoint'
]

corr_df = df[correlation_cols].dropna()
corr_matrix = corr_df.corr()


fig = px.imshow(
    corr_matrix,
    text_auto='.2f',
    color_continuous_scale='RdBu_r',
    zmin=-1, zmax=1,
    title='Correlation Matrix: Key Commercial Variables'
)
fig.update_layout(width=700, height=600)

fig.show()
```

### Interpretation and Business Insight

Estimated ownership ties most closely to review count and peak concurrent users. Price and a high positive ratio matter less. In other words, a game’s success rides on energy in its player community, not just on pricing strategy.

### Recommendation

That means you should push for community events, partnerships with creators, and campaigns that drive more players and more reviews. Those things fuel your growth engine.

## Q7. What ownership level should a new studio realistically target?

**Business value:** A precise target helps the studio to manage expectations.


```python
ownership_percentiles = np.percentile(df['Estimated owners midpoint'].dropna(), [50, 75, 90, 95, 99])

owner_range_counts = df['Estimated owners'].value_counts(normalize=True).mul(100).rename('Percent').reset_index().rename(columns={'index': 'Owner Range'}).sort_values('Percent', ascending=False)
owner_range_counts.head(8)
```

```python
fig = px.bar(
    owner_range_counts.head(8),
    x='Percent',
    y='Estimated owners',
    orientation='h',
    text=owner_range_counts['Percent'].head(8).apply(lambda x: f'{x:.1f}%'),
    title='Distribution of Steam Games by Estimated Ownership Range',
    labels={'Percent': 'Percentage of Games (%)', 'Estimated owners': 'Ownership Range'}
)
fig.update_layout(yaxis={'categoryorder': 'total ascending'}, width=1000, height=600)
fig.show()
```

### Interpretation and Business Insight

Most games on Steam don’t become massive hits over 60% have fewer than 20,000 owners, and few ever cross 100,000. Blockbuster sales are rare.

### Recommendation

You need realistic targets. Structure your milestones around ownership tiers, and plan your finances with the expectation that you’ll hit solid not stratospheric numbers. Don’t let exceptions skew your strategy.
